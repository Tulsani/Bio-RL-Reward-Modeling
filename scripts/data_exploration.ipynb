```python
import pandas as pd

df = pd.read_csv("trajectories_train.csv")

reward_cols = [
    "next_6h_map_delta",
    "next_6h_lactate_delta",
    "next_6h_deterioration",
    "adverse_hypotension_next_6h",
    "adverse_fluid_overload_next_6h",
    "adverse_tachyarrhythmia_next_6h",
]

print(df[reward_cols].describe())

for col in reward_cols[2:]:
    print(
        col,
        df[col].value_counts(normalize=True, dropna=False)
    )
```

```python
reward_cols = [
    "next_6h_map_delta",
    "next_6h_lactate_delta",
    "next_6h_deterioration",
    "adverse_hypotension_next_6h",
    "adverse_fluid_overload_next_6h",
    "adverse_tachyarrhythmia_next_6h",
]

print(
    df.groupby("observed_clinician_action")[reward_cols]
      .mean()
      .round(3)
)

print(
    df.groupby("observed_clinician_action")[
        ["map_mm_hg", "lactate_mmol_l", "heart_rate",
         "urine_output_6h_ml", "vasopressor_active_pre_action"]
    ]
    .mean()
    .round(2)
)

low_map = df["map_mm_hg"] < 65

print(
    df.assign(low_map=low_map)
      .groupby(["low_map", "observed_clinician_action"])[reward_cols]
      .mean()
      .round(3)
)
```

```python
print(df[reward_cols].corr().round(2))
```

```python
cols = [
    "lactate_mmol_l",
    "urine_output_6h_ml",
    "creatinine_mg_dl",
]

for col in cols:
    print(
        df.groupby("observed_clinician_action")[col]
          .apply(lambda x: x.isna().mean())
    )
```
